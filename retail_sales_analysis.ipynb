# Real-World Data Project – Retail Sales Analysis

Analyze a practice retail sales dataset using Python.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("retail_sales_dataset.csv")
df["Date"] = pd.to_datetime(df["Date"])
df.head()
```

## 1. Dataset Inspection and Data Quality

```python
print("Shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nData types:")
print(df.dtypes)
```

## 2. Statistical Summary

```python
df.describe(include='all').T
```

## 3. Monthly Sales Trend

```python
monthly_sales = df.set_index("Date")["Sales"].resample("ME").sum()
plt.figure(figsize=(10,5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 4. Sales by Category

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,5))
category_sales.plot(kind="bar")
plt.title("Sales by Product Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

## 5. Sales by Region

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,5))
region_sales.plot(kind="bar")
plt.title("Sales by Region")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

## 6. Payment Method Analysis

```python
payment_counts = df["Payment_Method"].value_counts()
plt.figure(figsize=(8,5))
payment_counts.plot(kind="bar")
plt.title("Transactions by Payment Method")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```

## 7. Quantity vs Sales

```python
plt.figure(figsize=(9,6))
sns.scatterplot(data=df, x="Quantity", y="Sales", hue="Category")
plt.title("Quantity vs Sales")
plt.tight_layout()
plt.show()
```

## 8. Correlation Analysis

```python
numeric_cols = ["Quantity", "Unit_Price", "Discount", "Sales"]
corr = df[numeric_cols].corr()
plt.figure(figsize=(7,5))
sns.heatmap(corr, annot=True, fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
print(corr["Sales"].sort_values(ascending=False))
```

## 9. Key Findings and Conclusion

- Monthly sales show changes across the year.
- Category and regional summaries show which groups contribute more sales.
- Payment-method counts show transaction preferences.
- Quantity and sales can be compared using the scatter plot.
- Correlation shows statistical association, not causation.
